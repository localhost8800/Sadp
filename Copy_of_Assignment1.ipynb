```python
import numpy as np
import matplotlib.pyplot as plt

# Input values
x = np.linspace(-10, 10, 200)

# Activation Functions

# a) Linear
linear = x

# b) Sigmoid
sigmoid = 1 / (1 + np.exp(-x))

# c) Tanh
tanh = np.tanh(x)

# d) ReLU
relu = np.maximum(0, x)

# e) Softmax
# Softmax is usually applied to a vector
softmax = np.exp(x - np.max(x))
softmax = softmax / np.sum(softmax)

# Plotting
plt.figure(figsize=(12, 8))

# Linear
plt.subplot(2, 3, 1)
plt.plot(x, linear, color='blue')
plt.title("Linear")
plt.grid(True)

# Sigmoid
plt.subplot(2, 3, 2)
plt.plot(x, sigmoid, color='green')
plt.title("Sigmoid")
plt.grid(True)

# Tanh
plt.subplot(2, 3, 3)
plt.plot(x, tanh, color='red')
plt.title("Tanh")
plt.grid(True)

# ReLU
plt.subplot(2, 3, 4)
plt.plot(x, relu, color='purple')
plt.title("ReLU")
plt.grid(True)

# Softmax
plt.subplot(2, 3, 5)
plt.plot(x, softmax, color='orange')
plt.title("Softmax")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.utils import to_categorical

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize pixel values (0-255 to 0-1)
x_train = x_train / 255.0
x_test = x_test / 255.0

# One-hot encode the labels
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# Build Single Layer Perceptron model
model = Sequential([
    Flatten(input_shape=(28, 28)),          # Convert 28x28 image into 784 features
    Dense(10, activation='softmax')         # Single output layer
])

# Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Train the model
model.fit(x_train, y_train,
          epochs=10,
          batch_size=32,
          validation_split=0.2)

# Evaluate the model
loss, accuracy = model.evaluate(x_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy * 100, "%")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Input values
x = np.linspace(-3, 3, 100)

# Hard Tanh
def hard_tanh(x):
    return np.clip(x, -1, 1)

# Leaky ReLU
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

# ELU
def elu(x, alpha=1.0):
    return np.where(x > 0, x, alpha * (np.exp(x) - 1))

# Plot graphs
plt.figure(figsize=(12,4))

# Hard Tanh
plt.subplot(1,3,1)
plt.plot(x, hard_tanh(x), color='blue')
plt.title("Hard Tanh")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)

# Leaky ReLU
plt.subplot(1,3,2)
plt.plot(x, leaky_relu(x), color='green')
plt.title("Leaky ReLU")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)

# ELU
plt.subplot(1,3,3)
plt.plot(x, elu(x), color='red')
plt.title("ELU")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Input values
x = np.linspace(-5, 5, 100)

# Parametric ReLU Function
def prelu(x, alpha=0.25):
    return np.where(x > 0, x, alpha * x)

# Apply PReLU
y = prelu(x)

# Plot
plt.plot(x, y, color='purple', label='PReLU')
plt.title("Parametric ReLU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.legend()
plt.show()
```
